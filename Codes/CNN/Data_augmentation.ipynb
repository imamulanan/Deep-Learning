```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
import matplotlib.pyplot as plt

```

```python
# Load image with target size
img = load_img('/content/drive/MyDrive/Colab Notebooks/Data Augmentation/cat.6611.jpg', target_size=(200, 200))

# Display
plt.imshow(img)
plt.axis('off')
plt.show()

# Convert to array if you need it for a model
img_array = img_to_array(img)
img_array = img_array / 255.0  # normalize
```

```python
type(img)
```

```python
# Cell 6: Create ImageDataGenerator for data augmentation
datagen = ImageDataGenerator(
    rotation_range=30,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
)

```

```python
# Cell 7: Convert image to numpy array
img = img_to_array(img)
```

```python
# Cell 8: Check the type of the image
type(img)
```

```python
# Cell 38: Check the shape of the image
img.shape
# Output: (200, 200, 3)
```

```python
# Reshape image to add batch dimension
input_batch = img.reshape(1,200,200,3)

```

```python
i = 0

# Generate augmented images
for output in datagen.flow(input_batch, batch_size=1, save_to_dir='/content/drive/MyDrive/Colab Notebooks/Data Augmentation/AUG'):
    i = i + 1
    if i == 10:
        break
```
